```python
# ==========================================
# PHASE 3.1 - IMPORT LIBRARIES
# ==========================================

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    GlobalAveragePooling2D
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    ModelCheckpoint,
    EarlyStopping,
    ReduceLROnPlateau
)

print("TensorFlow version:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
# ==========================================
# PHASE 3.2 - PROJECT SETTINGS
# ==========================================

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.20
SEED = 42

DATASET_PATH = "/content/dataset"

print("Image size:", IMAGE_SIZE)
print("Batch size:", BATCH_SIZE)
print("Validation split:", VALIDATION_SPLIT)
```

```python
# ==========================================
# PHASE 3.3 - CHECK DATASET
# ==========================================

authentic_path = os.path.join(
    DATASET_PATH,
    "authentic"
)

forged_path = os.path.join(
    DATASET_PATH,
    "forged"
)

print("Authentic folder exists:",
      os.path.exists(authentic_path))

print("Forged folder exists:",
      os.path.exists(forged_path))

print(
    "Authentic images:",
    len(os.listdir(authentic_path))
)

print(
    "Forged images:",
    len(os.listdir(forged_path))
)
```

```python
# ==========================================
# PHASE 3.4 - TRAINING DATA GENERATOR
# ==========================================

train_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0,

    # Small image transformations
    rotation_range=10,
    width_shift_range=0.10,
    height_shift_range=0.10,
    zoom_range=0.10,

    horizontal_flip=True,

    # 20% of images are used for validation
    validation_split=VALIDATION_SPLIT
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,

    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,

    class_mode="binary",

    subset="training",

    shuffle=True,
    seed=SEED
)
```

```python
# ==========================================
# PHASE 3.5 - VALIDATION DATA GENERATOR
# ==========================================

validation_generator = train_datagen.flow_from_directory(
    DATASET_PATH,

    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,

    class_mode="binary",

    subset="validation",

    shuffle=False,
    seed=SEED
)
```

```python
# ==========================================
# PHASE 3.6 - CHECK CLASS LABELS
# ==========================================

print("Class mapping:")
print(train_generator.class_indices)
```

```python
# ==========================================
# PHASE 3.7 - DISPLAY SAMPLE IMAGES
# ==========================================

images, labels = next(train_generator)

plt.figure(figsize=(12, 8))

for i in range(8):

    plt.subplot(2, 4, i + 1)

    plt.imshow(images[i])

    if labels[i] == 0:
        title = "Authentic"
    else:
        title = "Forged"

    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# PHASE 3.8 - LOAD EFFICIENTNETB0
# ==========================================

base_model = EfficientNetB0(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

print("EfficientNetB0 loaded successfully.")
```

```python
# ==========================================
# PHASE 3.9 - FREEZE PRETRAINED MODEL
# ==========================================

base_model.trainable = False

print("Base model frozen.")
```

```python
# ==========================================
# PHASE 3.10 - BUILD CLASSIFICATION HEAD
# ==========================================

x = base_model.output

# Convert feature maps into one feature vector
x = GlobalAveragePooling2D()(x)

# Reduce overfitting
x = Dropout(0.30)(x)

# Binary output:
# 0 = Authentic
# 1 = Forged
output = Dense(
    1,
    activation="sigmoid"
)(x)

model = Model(
    inputs=base_model.input,
    outputs=output
)

print("Model created.")
```

```python
# ==========================================
# PHASE 3.11 - COMPILE MODEL
# ==========================================

model.compile(
    optimizer=Adam(
        learning_rate=0.0001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy"
    ]
)

print("Model compiled.")
```

```python
# ==========================================
# PHASE 3.12 - MODEL SUMMARY
# ==========================================

model.summary()
```

```python
# ==========================================
# PHASE 3.13 - TRAINING CALLBACKS
# ==========================================

checkpoint = ModelCheckpoint(
    "/content/best_forgery_model.keras",

    monitor="val_accuracy",

    save_best_only=True,

    mode="max",

    verbose=1
)

early_stopping = EarlyStopping(
    monitor="val_loss",

    patience=5,

    restore_best_weights=True,

    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",

    factor=0.5,

    patience=2,

    min_lr=1e-7,

    verbose=1
)

print("Callbacks ready.")
```

```python
# ==========================================
# PHASE 3.14 - FINAL CHECK
# ==========================================

print("================================")
print("PHASE 3 COMPLETE")
print("================================")

print("Training images:",
      train_generator.samples)

print("Validation images:",
      validation_generator.samples)

print("Classes:",
      train_generator.class_indices)

print("Image size:",
      IMAGE_SIZE)

print("Batch size:",
      BATCH_SIZE)

print("Model:",
      "EfficientNetB0")

print("Optimizer:",
      "Adam")

print("Loss:",
      "Binary Crossentropy")

print("================================")
print("READY FOR TRAINING")
print("================================")
```

```python
# ============================================================
# PHASE 4A.1 - IMPORTS
# ============================================================

import os
import shutil
import random
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers
from tensorflow.keras import Model

from tensorflow.keras.applications import EfficientNetB0

from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    ModelCheckpoint,
    ReduceLROnPlateau
)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
# ============================================================
# PHASE 4A.2 - DATASET PATHS
# ============================================================

CASIA_PATH = "/content/CASIA2"

AUTHENTIC_PATH = os.path.join(
    CASIA_PATH,
    "Au"
)

FORGED_PATH = os.path.join(
    CASIA_PATH,
    "Tp"
)

print("Authentic path:")
print(AUTHENTIC_PATH)

print("\nForged path:")
print(FORGED_PATH)

print("\nAuthentic folder exists:",
      os.path.exists(AUTHENTIC_PATH))

print("Forged folder exists:",
      os.path.exists(FORGED_PATH))
```

```python
# ============================================================
# PHASE 4A.3 - COUNT DATASET FILES
# ============================================================

authentic_files = [
    f for f in os.listdir(AUTHENTIC_PATH)
    if os.path.isfile(
        os.path.join(AUTHENTIC_PATH, f)
    )
]

forged_files = [
    f for f in os.listdir(FORGED_PATH)
    if os.path.isfile(
        os.path.join(FORGED_PATH, f)
    )
]

print("Authentic files:", len(authentic_files))
print("Forged files:", len(forged_files))
```

```python
# ============================================================
# PHASE 4A.4 - CHECK IMAGE EXTENSIONS
# ============================================================

from collections import Counter

auth_extensions = Counter(
    os.path.splitext(f)[1].lower()
    for f in authentic_files
)

forged_extensions = Counter(
    os.path.splitext(f)[1].lower()
    for f in forged_files
)

print("Authentic extensions:")
print(auth_extensions)

print("\nForged extensions:")
print(forged_extensions)
```

```python
# ============================================================
# PHASE 4A.5 - VISUAL DATA CHECK
# ============================================================

import cv2

random.seed(42)

auth_samples = random.sample(
    authentic_files,
    min(4, len(authentic_files))
)

forged_samples = random.sample(
    forged_files,
    min(4, len(forged_files))
)

plt.figure(figsize=(14, 7))

# Authentic images
for i, filename in enumerate(auth_samples):

    path = os.path.join(
        AUTHENTIC_PATH,
        filename
    )

    image = cv2.imread(path)

    if image is None:
        continue

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    plt.subplot(2, 4, i + 1)
    plt.imshow(image)
    plt.title("Authentic")
    plt.axis("off")

# Forged images
for i, filename in enumerate(forged_samples):

    path = os.path.join(
        FORGED_PATH,
        filename
    )

    image = cv2.imread(path)

    if image is None:
        continue

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    plt.subplot(2, 4, i + 5)
    plt.imshow(image)
    plt.title("Forged")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 4A.6 - CREATE BALANCED DATASET
# ============================================================

BALANCED_PATH = "/content/balanced_dataset"

BALANCED_AUTH = os.path.join(
    BALANCED_PATH,
    "authentic"
)

BALANCED_FORGED = os.path.join(
    BALANCED_PATH,
    "forged"
)

# Remove previous balanced dataset if it exists
if os.path.exists(BALANCED_PATH):
    shutil.rmtree(BALANCED_PATH)

os.makedirs(BALANCED_AUTH)
os.makedirs(BALANCED_FORGED)

# Number of images to use
BALANCED_COUNT = min(
    len(authentic_files),
    len(forged_files)
)

print("Images per class:", BALANCED_COUNT)
```

```python
# ============================================================
# PHASE 4A.7 - COPY AUTHENTIC IMAGES
# ============================================================

random.seed(42)

selected_authentic = random.sample(
    authentic_files,
    BALANCED_COUNT
)

for filename in selected_authentic:

    source = os.path.join(
        AUTHENTIC_PATH,
        filename
    )

    destination = os.path.join(
        BALANCED_AUTH,
        filename
    )

    shutil.copy2(
        source,
        destination
    )

print(
    "Authentic images copied:",
    len(os.listdir(BALANCED_AUTH))
)
```

```python
# ============================================================
# PHASE 4A.8 - COPY FORGED IMAGES
# ============================================================

random.seed(42)

selected_forged = random.sample(
    forged_files,
    BALANCED_COUNT
)

for filename in selected_forged:

    source = os.path.join(
        FORGED_PATH,
        filename
    )

    destination = os.path.join(
        BALANCED_FORGED,
        filename
    )

    shutil.copy2(
        source,
        destination
    )

print(
    "Forged images copied:",
    len(os.listdir(BALANCED_FORGED))
)
```

```python
# ============================================================
# PHASE 4A.9 - VERIFY BALANCED DATASET
# ============================================================

print("================================")
print("BALANCED DATASET")
print("================================")

print(
    "Authentic:",
    len(os.listdir(BALANCED_AUTH))
)

print(
    "Forged:",
    len(os.listdir(BALANCED_FORGED)
)

)
```

```python
# ============================================================
# PHASE 4A.10 - CLEAN DATA GENERATORS
# ============================================================

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.20

# Very light augmentation
train_datagen = ImageDataGenerator(
    rotation_range=5,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.05,

    validation_split=VALIDATION_SPLIT
)

# No augmentation for validation
validation_datagen = ImageDataGenerator(
    validation_split=VALIDATION_SPLIT
)

train_generator = train_datagen.flow_from_directory(
    BALANCED_PATH,

    target_size=IMAGE_SIZE,

    batch_size=BATCH_SIZE,

    class_mode="binary",

    subset="training",

    shuffle=True,

    seed=42
)

validation_generator = validation_datagen.flow_from_directory(
    BALANCED_PATH,

    target_size=IMAGE_SIZE,

    batch_size=BATCH_SIZE,

    class_mode="binary",

    subset="validation",

    shuffle=False,

    seed=42
)

print("\n================================")
print("GENERATOR CHECK")
print("================================")

print(
    "Training images:",
    train_generator.samples
)

print(
    "Validation images:",
    validation_generator.samples
)

print(
    "Classes:",
    train_generator.class_indices
)
```

```python
# ============================================================
# PHASE 4A.11 - GENERATOR VISUAL CHECK
# ============================================================

images, labels = next(train_generator)

plt.figure(figsize=(14, 8))

for i in range(8):

    plt.subplot(2, 4, i + 1)

    plt.imshow(
        images[i].astype("uint8")
    )

    if labels[i] == 0:
        label = "Authentic"
    else:
        label = "Forged"

    plt.title(label)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 4A.12 - FRESH EFFICIENTNETB0
# ============================================================

base_model = EfficientNetB0(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze initially
base_model.trainable = False

print("EfficientNetB0 loaded.")
print("Base model frozen.")
```

```python
# ============================================================
# PHASE 4A.13 - CLASSIFICATION MODEL
# ============================================================

inputs = tf.keras.Input(
    shape=(224, 224, 3)
)

# IMPORTANT:
# training=False keeps BatchNormalization layers
# in inference mode during transfer learning/fine-tuning.
x = base_model(
    inputs,
    training=False
)

x = tf.keras.layers.GlobalAveragePooling2D()(x)

x = tf.keras.layers.Dropout(
    0.30
)(x)

x = tf.keras.layers.Dense(
    128,
    activation="relu"
)(x)

x = tf.keras.layers.Dropout(
    0.30
)(x)

outputs = tf.keras.layers.Dense(
    1,
    activation="sigmoid"
)(x)

model = tf.keras.Model(
    inputs,
    outputs
)

print("Model created.")
```

```python
# ============================================================
# PHASE 4A.14 - COMPILE
# ============================================================

model.compile(
    optimizer=Adam(
        learning_rate=0.0001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy"
    ]
)

print("Model compiled.")
```

```python
# ============================================================
# PHASE 4A.15 - MODEL SUMMARY
# ============================================================

model.summary()
```

```python
# ============================================================
# PHASE 4A.16 - CALLBACKS
# ============================================================

checkpoint_stage1 = ModelCheckpoint(
    "/content/clean_stage1_best.keras",

    monitor="val_accuracy",

    save_best_only=True,

    mode="max",

    verbose=1
)

reduce_lr_stage1 = ReduceLROnPlateau(
    monitor="val_loss",

    factor=0.5,

    patience=2,

    min_lr=1e-7,

    verbose=1
)

print("Callbacks ready.")
```

```python
# ============================================================
# PHASE 4A.17 - STAGE 1 TRAINING
# ============================================================

STAGE1_EPOCHS = 15

print("======================================")
print("STAGE 1 TRAINING")
print("======================================")

history_stage1 = model.fit(

    train_generator,

    validation_data=validation_generator,

    epochs=STAGE1_EPOCHS,

    callbacks=[
        checkpoint_stage1,
        reduce_lr_stage1
    ],

    verbose=1
)

print("\nStage 1 training completed.")
```

```python
# ============================================================
# PHASE 4A.18 - LOAD BEST MODEL
# ============================================================

best_stage1_model = tf.keras.models.load_model(
    "/content/clean_stage1_best.keras"
)

print("Best Stage 1 model loaded.")
```

```python
# ============================================================
# PHASE 4A.19 - STAGE 1 EVALUATION
# ============================================================

validation_generator.reset()

stage1_result = best_stage1_model.evaluate(
    validation_generator,
    verbose=1
)

stage1_loss = stage1_result[0]
stage1_accuracy = stage1_result[1]

print("\n======================================")
print("STAGE 1 RESULT")
print("======================================")

print(
    "Validation Loss:",
    round(stage1_loss, 4)
)

print(
    "Validation Accuracy:",
    round(stage1_accuracy * 100, 2),
    "%"
)
```

```python
# ============================================================
# PHASE 4A.20 - FINE-TUNING SETUP
# ============================================================

model = best_stage1_model

# Find the EfficientNet base model
base_model = None

for layer in model.layers:

    if isinstance(
        layer,
        tf.keras.Model
    ) and "efficientnet" in layer.name.lower():

        base_model = layer
        break

if base_model is None:

    print("EfficientNet base model not found.")

    print("\nModel layers:")

    for layer in model.layers:
        print(
            layer.name,
            type(layer)
        )

    raise ValueError(
        "EfficientNet base model could not be found."
    )

print(
    "EfficientNet base model:",
    base_model.name
)
```

```python
# ============================================================
# PHASE 4A.21 - UNFREEZE UPPER LAYERS
# ============================================================

base_model.trainable = True

# Freeze most layers
for layer in base_model.layers[:-40]:

    layer.trainable = False

# Unfreeze final 40 layers,
# except BatchNormalization layers.
for layer in base_model.layers[-40:]:

    if isinstance(
        layer,
        tf.keras.layers.BatchNormalization
    ):

        layer.trainable = False

    else:

        layer.trainable = True

trainable_layers = 0
frozen_layers = 0

for layer in base_model.layers:

    if layer.trainable:
        trainable_layers += 1
    else:
        frozen_layers += 1

print("================================")
print("FINE-TUNING SETUP")
print("================================")

print(
    "Trainable EfficientNet layers:",
    trainable_layers
)

print(
    "Frozen EfficientNet layers:",
    frozen_layers
)
```

```python
# ============================================================
# PHASE 4A.22 - RECOMPILE FOR FINE-TUNING
# ============================================================

model.compile(
    optimizer=Adam(
        learning_rate=1e-5
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy"
    ]
)

print(
    "Fine-tuning learning rate: 0.00001"
)
```

```python
# ============================================================
# PHASE 4A.23 - FINE-TUNING CALLBACKS
# ============================================================

checkpoint_stage2 = ModelCheckpoint(
    "/content/clean_final_forgery_model.keras",

    monitor="val_accuracy",

    save_best_only=True,

    mode="max",

    verbose=1
)

reduce_lr_stage2 = ReduceLROnPlateau(
    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=1e-8,

    verbose=1
)

print("Fine-tuning callbacks ready.")
```

```python
# ============================================================
# PHASE 4A.24 - FINE-TUNING
# ============================================================

STAGE2_EPOCHS = 15

print("======================================")
print("STAGE 2 - FINE-TUNING")
print("======================================")

history_stage2 = model.fit(

    train_generator,

    validation_data=validation_generator,

    epochs=STAGE2_EPOCHS,

    callbacks=[
        checkpoint_stage2,
        reduce_lr_stage2
    ],

    verbose=1
)

print("\nFine-tuning completed.")
```

```python
# ============================================================
# PHASE 4A.25 - LOAD FINAL MODEL
# ============================================================

final_model = tf.keras.models.load_model(
    "/content/clean_final_forgery_model.keras"
)

print("Final best model loaded.")
```

```python
# ============================================================
# PHASE 4A.26 - FINAL EVALUATION
# ============================================================

validation_generator.reset()

final_result = final_model.evaluate(
    validation_generator,
    verbose=1
)

final_loss = final_result[0]
final_accuracy = final_result[1]

print("\n======================================")
print("FINAL CLEAN TRAINING RESULT")
print("======================================")

print(
    "Validation Loss:",
    round(final_loss, 4)
)

print(
    "Validation Accuracy:",
    round(final_accuracy * 100, 2),
    "%"
)
```

```python
# ============================================================
# PHASE 12 - FINAL CORRECTED GRAD-CAM
# ============================================================

import os
import cv2
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# ============================================================
# 1. LOAD MODEL
# ============================================================

MODEL_PATH = "/content/clean_final_forgery_model.keras"

model = tf.keras.models.load_model(
    MODEL_PATH
)

print("Model loaded successfully.")


# ============================================================
# 2. FIND EFFICIENTNET BASE MODEL
# ============================================================

base_model = None

for layer in model.layers:

    if isinstance(layer, tf.keras.Model):

        if "efficientnet" in layer.name.lower():

            base_model = layer
            break


if base_model is None:

    raise ValueError(
        "EfficientNet base model was not found."
    )


print("EfficientNet base model:")
print(base_model.name)


# ============================================================
# 3. FIND LAST CONVOLUTIONAL LAYER
# ============================================================

last_conv_layer = None

for layer in reversed(base_model.layers):

    if isinstance(
        layer,
        (
            tf.keras.layers.Conv2D,
            tf.keras.layers.DepthwiseConv2D
        )
    ):

        last_conv_layer = layer
        break


if last_conv_layer is None:

    # fallback: any layer producing 4D output
    for layer in reversed(base_model.layers):

        try:

            if len(layer.output.shape) == 4:

                last_conv_layer = layer
                break

        except Exception:

            pass


if last_conv_layer is None:

    raise ValueError(
        "Could not find convolutional layer."
    )


print("\nGrad-CAM layer:")
print(last_conv_layer.name)

print(
    "Output shape:",
    last_conv_layer.output.shape
)


# ============================================================
# 4. IDENTIFY CLASSIFIER LAYERS
# ============================================================
#
# The outer model looks approximately like:
#
# Input
#   ↓
# EfficientNet
#   ↓
# GlobalAveragePooling
#   ↓
# Dropout
#   ↓
# Dense
#   ↓
# Dropout
#   ↓
# Dense(sigmoid)
#
# We reuse the classifier layers after EfficientNet.

base_index = None

for i, layer in enumerate(model.layers):

    if layer is base_model:

        base_index = i
        break


if base_index is None:

    raise ValueError(
        "Could not locate EfficientNet inside model."
    )


classifier_layers = model.layers[
    base_index + 1:
]


print("\nClassifier layers:")

for layer in classifier_layers:

    print(
        layer.name,
        type(layer).__name__
    )


# ============================================================
# 5. BUILD GRADIENT MODEL
# ============================================================
#
# IMPORTANT:
#
# We construct everything from the SAME EfficientNet graph.
#
# This fixes:
#
# KeyError
# gradients are None
# input structure mismatch
#
# ============================================================

grad_model = tf.keras.Model(

    inputs=base_model.input,

    outputs=[
        last_conv_layer.output,
        base_model.output
    ]

)


print("\nGradient feature model created.")


# ============================================================
# 6. IMAGE PREPARATION
# ============================================================

IMAGE_SIZE = (224, 224)


def prepare_image(image_path):

    image = cv2.imread(
        image_path
    )

    if image is None:

        raise ValueError(
            "Could not read image:\n"
            + image_path
        )

    # BGR → RGB
    image_rgb = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Resize
    resized = cv2.resize(
        image_rgb,
        IMAGE_SIZE
    )

    # EfficientNet model accepts pixel values.
    # Do NOT add 1/255 here because the Keras
    # EfficientNet implementation handles its
    # preprocessing internally.

    tensor = resized.astype(
        np.float32
    )

    # Add batch dimension
    tensor = np.expand_dims(
        tensor,
        axis=0
    )

    return (
        image_rgb,
        tensor
    )


# ============================================================
# 7. GRAD-CAM FUNCTION
# ============================================================

def generate_gradcam(
    image_tensor
):

    # --------------------------------------------------------
    # Gradient calculation
    # --------------------------------------------------------

    with tf.GradientTape() as tape:

        # Get:
        #
        # 1. last convolutional feature maps
        # 2. final EfficientNet feature representation

        conv_outputs, base_outputs = grad_model(
            image_tensor,
            training=False
        )

        # IMPORTANT:
        # Watch the convolution output directly.
        tape.watch(
            conv_outputs
        )

        # ----------------------------------------------------
        # Run the classifier
        # ----------------------------------------------------

        x = base_outputs

        for layer in classifier_layers:

            x = layer(
                x,
                training=False
            )

        predictions = x

        # Forged probability
        forged_score = predictions[:, 0]


    # --------------------------------------------------------
    # Calculate gradients
    # --------------------------------------------------------

    gradients = tape.gradient(
        forged_score,
        conv_outputs
    )


    # Safety check
    if gradients is None:

        raise ValueError(
            "Gradients are still None. "
            "The selected layer is not connected "
            "to the classifier."
        )


    # --------------------------------------------------------
    # Global average pooling of gradients
    # --------------------------------------------------------

    pooled_gradients = tf.reduce_mean(
        gradients,
        axis=(1, 2)
    )


    # Remove batch dimension
    conv_outputs = conv_outputs[0]

    pooled_gradients = pooled_gradients[0]


    # --------------------------------------------------------
    # Weight feature maps
    # --------------------------------------------------------

    weighted_features = (
        conv_outputs *
        pooled_gradients
    )


    # Sum across channels
    heatmap = tf.reduce_sum(
        weighted_features,
        axis=-1
    )


    # ReLU
    heatmap = tf.maximum(
        heatmap,
        0
    )


    # Normalize
    max_heat = tf.reduce_max(
        heatmap
    )

    heatmap = tf.where(
        max_heat > 0,
        heatmap / max_heat,
        heatmap
    )


    return (
        heatmap.numpy(),
        float(
            forged_score.numpy()[0]
        )
    )


# ============================================================
# 8. CREATE HEATMAP OVERLAY
# ============================================================

def create_overlay(
    original_image,
    heatmap
):

    # Convert heatmap to uint8
    heatmap_uint8 = np.uint8(
        heatmap * 255
    )

    # Resize
    heatmap_uint8 = cv2.resize(
        heatmap_uint8,
        (
            original_image.shape[1],
            original_image.shape[0]
        )
    )

    # Color map
    colored = cv2.applyColorMap(
        heatmap_uint8,
        cv2.COLORMAP_JET
    )

    # RGB → BGR
    original_bgr = cv2.cvtColor(
        original_image,
        cv2.COLOR_RGB2BGR
    )

    # Overlay
    overlay = cv2.addWeighted(
        original_bgr,
        0.60,
        colored,
        0.40,
        0
    )

    # BGR → RGB
    overlay = cv2.cvtColor(
        overlay,
        cv2.COLOR_BGR2RGB
    )

    return (
        heatmap_uint8,
        overlay
    )


# ============================================================
# 9. FIND SUSPICIOUS REGION
# ============================================================

def find_suspicious_region(
    original_image,
    heatmap,
    threshold=0.55
):

    # Normalize
    if heatmap.max() > 0:

        heatmap_normalized = (
            heatmap /
            heatmap.max()
        )

    else:

        heatmap_normalized = heatmap


    # Resize
    heatmap_resized = cv2.resize(
        heatmap_normalized,
        (
            original_image.shape[1],
            original_image.shape[0]
        )
    )


    # Threshold
    mask = (
        heatmap_resized >= threshold
    ).astype(
        np.uint8
    ) * 255


    # Remove noise
    kernel = np.ones(
        (7, 7),
        np.uint8
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel
    )


    # Find contours
    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # Find largest region
    best_contour = None
    best_area = 0


    for contour in contours:

        area = cv2.contourArea(
            contour
        )

        if area > best_area:

            best_area = area

            best_contour = contour


    # Create output
    boxed = original_image.copy()


    if best_contour is not None:

        x, y, w, h = cv2.boundingRect(
            best_contour
        )

        cv2.rectangle(
            boxed,

            (x, y),

            (x + w, y + h),

            (255, 0, 0),

            4
        )

        cv2.putText(
            boxed,

            "Suspicious Region",

            (
                x,
                max(
                    30,
                    y - 10
                )
            ),

            cv2.FONT_HERSHEY_SIMPLEX,

            0.8,

            (255, 0, 0),

            2,

            cv2.LINE_AA
        )

        bounding_box = (
            x,
            y,
            w,
            h
        )

    else:

        bounding_box = None


    return (
        mask,
        boxed,
        bounding_box,
        best_area
    )


# ============================================================
# 10. COMPLETE IMAGE ANALYZER
# ============================================================

def analyze_forgery(
    image_path,
    threshold=0.55
):

    print("\n")
    print("=" * 60)
    print("ANALYZING")
    print("=" * 60)

    print(image_path)


    # Load image
    original_image, image_tensor = (
        prepare_image(
            image_path
        )
    )


    # Generate Grad-CAM
    heatmap, score = generate_gradcam(
        image_tensor
    )


    # Classification
    if score >= 0.50:

        label = "FORGED"

    else:

        label = "AUTHENTIC"


    # Overlay
    heatmap_image, overlay = (
        create_overlay(
            original_image,
            heatmap
        )
    )


    # Suspicious region
    mask, boxed_image, bounding_box, area = (
        find_suspicious_region(
            original_image,
            heatmap,
            threshold
        )
    )


    # Print result
    print("\n")
    print("=" * 60)
    print("RESULT")
    print("=" * 60)

    print(
        "Prediction:",
        label
    )

    print(
        f"Forgery Score: {score * 100:.2f}%"
    )

    print(
        f"Heatmap threshold: {threshold}"
    )


    if bounding_box is not None:

        print(
            "Suspicious bounding box:",
            bounding_box
        )

        print(
            f"Suspicious area: {area:.0f} pixels"
        )

    else:

        print(
            "Suspicious region: Not detected"
        )


    return {

        "original":
            original_image,

        "heatmap":
            heatmap,

        "heatmap_image":
            heatmap_image,

        "overlay":
            overlay,

        "mask":
            mask,

        "boxed_image":
            boxed_image,

        "score":
            score,

        "label":
            label,

        "bounding_box":
            bounding_box,

        "area":
            area
    }


# ============================================================
# 11. TEST YOUR FORGED IMAGE
# ============================================================

test_forged_image = (
    "/content/balanced_dataset/forged/"
    "Tp_D_NRN_S_N_art00013_art00082_11816.jpg"
)


result_forged = analyze_forgery(
    test_forged_image
)


# ============================================================
# 12. DISPLAY RESULTS
# ============================================================

plt.figure(
    figsize=(18, 10)
)


# Original
plt.subplot(
    2,
    3,
    1
)

plt.imshow(
    result_forged["original"]
)

plt.title(
    "Original Image"
)

plt.axis("off")


# Heatmap
plt.subplot(
    2,
    3,
    2
)

plt.imshow(
    result_forged["heatmap"],
    cmap="jet"
)

plt.title(
    "Grad-CAM Heatmap"
)

plt.axis("off")


# Overlay
plt.subplot(
    2,
    3,
    3
)

plt.imshow(
    result_forged["overlay"]
)

plt.title(
    "Heatmap Overlay"
)

plt.axis("off")


# Suspicious region
plt.subplot(
    2,
    3,
    4
)

plt.imshow(
    result_forged["boxed_image"]
)

plt.title(
    "Suspicious Region"
)

plt.axis("off")


# Mask
plt.subplot(
    2,
    3,
    5
)

plt.imshow(
    result_forged["mask"],
    cmap="gray"
)

plt.title(
    "Suspicious Mask"
)

plt.axis("off")


# Score
plt.subplot(
    2,
    3,
    6
)

plt.axis("off")

plt.text(
    0.5,
    0.65,
    result_forged["label"],
    ha="center",
    va="center",
    fontsize=25
)

plt.text(
    0.5,
    0.40,
    f"Forgery Score\n"
    f"{result_forged['score'] * 100:.2f}%",
    ha="center",
    va="center",
    fontsize=20
)


plt.tight_layout()

plt.show()


# ============================================================
# 13. SAVE RESULTS
# ============================================================

RESULT_DIR = (
    "/content/phase12_results"
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)


# Original
cv2.imwrite(
    f"{RESULT_DIR}/original.png",

    cv2.cvtColor(
        result_forged["original"],
        cv2.COLOR_RGB2BGR
    )
)


# Heatmap
cv2.imwrite(
    f"{RESULT_DIR}/heatmap.png",

    result_forged["heatmap_image"]
)


# Overlay
cv2.imwrite(
    f"{RESULT_DIR}/overlay.png",

    cv2.cvtColor(
        result_forged["overlay"],
        cv2.COLOR_RGB2BGR
    )
)


# Suspicious region
cv2.imwrite(
    f"{RESULT_DIR}/suspicious_region.png",

    cv2.cvtColor(
        result_forged["boxed_image"],
        cv2.COLOR_RGB2BGR
    )
)


# Mask
cv2.imwrite(
    f"{RESULT_DIR}/mask.png",

    result_forged["mask"]
)


print("\n")
print("=" * 60)
print("PHASE 12 COMPLETED SUCCESSFULLY")
print("=" * 60)

print(
    "Results folder:"
)

print(
    RESULT_DIR
)

print("\nFiles:")

print("original.png")
print("heatmap.png")
print("overlay.png")
print("suspicious_region.png")
print("mask.png")
```

```python
# ============================================================
# COMPLETE EVALUATION
# ============================================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    ConfusionMatrixDisplay
)

# ------------------------------------------------------------
# LOAD TEST DATASET
# ------------------------------------------------------------

test_dataset = tf.keras.utils.image_dataset_from_directory(
    "/content/balanced_dataset",
    labels="inferred",
    label_mode="binary",
    image_size=(224,224),
    batch_size=32,
    shuffle=False
)

# ------------------------------------------------------------
# PREDICTIONS
# ------------------------------------------------------------

y_true = []
y_scores = []

for images, labels in test_dataset:

    preds = model.predict(
        images,
        verbose=0
    ).flatten()

    y_scores.extend(preds)
    y_true.extend(labels.numpy().flatten())

y_true = np.array(y_true).astype(int)
y_scores = np.array(y_scores)

y_pred = (y_scores >= 0.5).astype(int)

# ------------------------------------------------------------
# METRICS
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

auc_score = roc_auc_score(
    y_true,
    y_scores
)

print("="*60)
print("MODEL EVALUATION")
print("="*60)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {auc_score:.4f}")

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\n")
print("="*60)
print("CLASSIFICATION REPORT")
print("="*60)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=[
            "Authentic",
            "Forged"
        ]
    )
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(
    y_true,
    y_pred
)

print("\n")
print("="*60)
print("CONFUSION MATRIX")
print("="*60)

print(cm)

# ------------------------------------------------------------
# CONFUSION MATRIX GRAPH
# ------------------------------------------------------------

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Authentic",
        "Forged"
    ]
)

disp.plot()

plt.title(
    "Confusion Matrix"
)

plt.show()

# ------------------------------------------------------------
# ROC CURVE
# ------------------------------------------------------------

fpr, tpr, _ = roc_curve(
    y_true,
    y_scores
)

plt.figure(figsize=(6,6))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {auc_score:.4f}"
)

plt.plot(
    [0,1],
    [0,1],
    "--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve"
)

plt.legend()

plt.show()

# ------------------------------------------------------------
# SCORE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

plt.hist(
    y_scores[y_true==0],
    bins=30,
    alpha=0.6,
    label="Authentic"
)

plt.hist(
    y_scores[y_true==1],
    bins=30,
    alpha=0.6,
    label="Forged"
)

plt.axvline(
    0.5,
    linestyle="--"
)

plt.xlabel(
    "Forgery Score"
)

plt.ylabel(
    "Count"
)

plt.title(
    "Prediction Score Distribution"
)

plt.legend()

plt.show()
```
